```python
#task2
def addition(x, y):
    return x + y

def subtraction(x, y):
    return x - y

def multiplication(x, y):
    return x * y
def division_with_if(x, y):
    if y == 0:
        return "Error: Division by zero"
    else:
        return x / y
def division_with_try_except(x, y):
    try:
        c = x / y
        return c
    except ZeroDivisionError:
        return "Error: Division by zero"

print("Addition:", addition(10, 5))
print("Subtraction:", subtraction(10, 5))
print("Multiplication:", multiplication(10, 5))
print("Division (If statement, y=0):", division_with_if(10, 0))
print("Division (Try-Except, y=0):", division_with_try_except(3, 0))
```
